```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def __str__(self):
        return f'{self.owner}: ${self.balance}'

acct = BankAccount('Mai', 100)
acct.deposit(50)
print(acct)     # Mai: $150
```

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance         # this calls the setter below

    @property
    def balance(self):                 # the getter
        return self.__balance

    @balance.setter
    def balance(self, amount):         # the setter (validates)
        if amount < 0:
            raise ValueError('balance cannot be negative')
        self.__balance = amount

acct = BankAccount('Mai', 100)
print(acct.balance)        # 100   <- looks like a normal attribute
acct.balance = 120         # runs the setter -> validated
print(acct.balance)        # 120
```

```python
What the @property decorator does
The @property decorator allows you to access a method as if it were a plain attribute (using simple dot notation without parentheses). Behind the scenes, it routes the request through a function, giving you clean syntax while keeping control over how data is read and modified.

2. Why __balance (double underscore) is hard to access
When an attribute starts with a double underscore, Python automatically renames it internally by prefixing it with the class name. Because the original variable name no longer exists in Python's memory under that exact name, trying to call it directly from outside the class causes Python to report that the attribute does not exist.

3. Difference between a getter/setter and a plain attribute

Plain Attribute: Opens direct access to data. Anyone can read or overwrite the value with anything—even invalid or dangerous data—without any restrictions.

Getter/Setter: Places a protective guard function around the data. Every time someone tries to view or change the value, the code checks the rules first, keeping the object safe from invalid states.

4. When to raise ValueError instead of return

Use raise ValueError: When the program receives input that breaks fundamental rules (like setting a negative bank balance). Raising an error immediately stops execution, preventing corrupted data from spreading through your program.

Use return: When a process completes its work normally and needs to hand a value back to the caller. Using return for bad input allows bad data to fail silently and create bigger bugs down the line.
```
